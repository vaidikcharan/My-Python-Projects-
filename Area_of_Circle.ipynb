Area = π × r²

where:

π (pi) is approximately 3.14159.
r is the radius of the circle.

```python
import math
print("Q1. Using math.pi: ")
r= 6
area= math.pi * (r ** 2 )
print(f"Area of circle is {area}")
```

```python
print("Q2. Using math.pow(): ")
r=6
area= math.pi * math.pow(r, 2)
print(f"Area of circle is {area}")
```

```python
import numpy as np
print("Q3. Using numpy.pi: ")
r=6
area= np.pi * (r ** 2)
print(f"Area of circle is {area}")
```

```python
print("Q4. Using hardcoded pi value:")
PI= 3.142
r=6
area= PI * (r ** 2)
print(f"Area of circle is {area}")
```
