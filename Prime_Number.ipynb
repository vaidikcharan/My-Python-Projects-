# Check Prime Number
Input: n = 29
Output: Prime Number

Input: n = 10
Output: Not a Prime Number

```python
print("Q1. Using flag variable: ")
n= 11
if(n <= 1):
  print(False)
else:
  prime = True
  for i in range(2, int(n ** 0.5)+1):
    if(n % i == 0):
      prime= False
      break
  print(prime)
```

```python
from sympy import isprime
print("Q2. Using isprime() method: ")
n = 11
print(isprime(n))
```

```python
print("Q3. Using Recursion: ")
from math import sqrt

def prime(n, i):
    if(i == 1):
        return True
    if(n % i == 0):
        return False
    return prime(n, i - 1)

n = 11
print(prime(n, int(sqrt(n))))
```

```python
print("Q4. Using function/recursion and input: ")
def prime(n, i):
    if(i == 1):
        return True
    if(n % i == 0):
        return False
    return prime(n, i - 1)
n= int(input("Enter a number: "))
print(prime(n, int(sqrt(n))))
```

```python
print("Q5. Using Loop: ")
def prime(n, i):
    if(i == 1):
        return True
    if(n % i == 0):
        return False
    return prime(n, i - 1)
n= int(input("Enter a number: "))
print(prime(n, int(sqrt(n))))
a= input("Continue Program? (y/n): ")
if(a == 'n'):
  print("Program Ended")
else:
  while(a == 'y'):
    n= int(input("Enter a number: "))
    print(prime(n, int(sqrt(n))))
    a= input("Continue Program? (y/n): ")
print("Program End.")
```
