**Python Program for Simple Interest**

Example: if P = 1000, T = 2 years, and R = 5%, the Simple Interest is 100.0.

simple interest formula : (P x T x R)/100

```python
print("Q1. Using Function: ")

def simple_interest(p, t, r):
    return (p * t * r) / 100

p = 1000
t = 2
r = 5
result = simple_interest(p, t, r)
print(result)
```

```python
print("Q2. Using function User input: ")

def simple_intrest(p, t, r):
  return (p * t * r)/100

p= int(input("Enter your Principle: "))
t= int(input("Enetr your time in years: "))
r= int(input("Enter your Rate: "))

result= simple_intrest(p, t, r)

print(f"Your simple intrest is: {result}")
```

```python
print("Q3. Using lambda function: ")
#A lambda function can perform the Simple Interest calculation in a single expression.

si= lambda p, t, r: (p * t * r)/100

p, t, r = 1000, 2, 5

result= si(p, t, r)
print(f"your Simple intrest is: {result}")
```

```python
print("Q4. Using list comprehension: ")
p, t, r= 1000, 2, 5

si= [p * t *r/100][0]
print(f"Your Simple intrest is: {si}")
```
