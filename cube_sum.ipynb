#  cube sum of first n natural numbers

nput: n = 5

Output: 225

Explanation:  1^3 + 2^3 + 3^3 + 4^3 + 5^3 = 225


```python
print("Q1. Using Mathematical Formula: ")
n= 7
csum= ((n * (n + 1)) // 2) ** 2
print(csum)
```

```python
print("Q2. Using Brute Force approach: ")
n = 7
csum = 0

for i in range(1, n + 1):
    csum += i ** 3

print(csum)
```

```python
import builtins
print("Q3. Using Generator Expression: ")
n = 7
csum = builtins.sum(i**3 for i in range(1, n + 1))

print(csum)
```

```python
print("Q4. Using for loop and break: ")
n = int(input("Enter a number: "))

if(n <= 1):
    print("Not a Prime Number")
else:
    prime = True

    for i in range(2, n):
        if n % i == 0:
            prime = False
            break

    if(prime):
        print("Prime Number")
    else:
        print("Not a Prime Number")
```

```python
print("Q5.Using for loop and else: ")
n = int(input("Enter a number: "))
if(n <= 1):
    print("Not a Prime Number")
else:
    for i in range(2, n):
        if(n % i == 0):
            print("Not a Prime Number")
            break
    else:
        print("Prime Number")
```
